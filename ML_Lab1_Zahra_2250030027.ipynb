```python
study_hours = [1, 2, 3, 4, 5, 6, 7]
marks = [50, 55, 60, 65, 72, 80, 88]

num_samples = len(marks)
print ("Number of samples: " + str(num_samples))

avg_hours = sum(study_hours) / float(num_samples)
print ("Average study hours: %.2f" % avg_hours)

avg_marks = sum(marks) / float(num_samples)
print ("Average mark: {:.2f}".format(avg_marks))


highest_mark = max(marks)
lowest_mark = min(marks)
print ("Highest mark: " + str(highest_mark))
print ("Lowest mark: " + str(lowest_mark))

for i in range(num_samples):
    if marks[i] >= 60:
        status = "Pass"
    else:
        status = "Fail"
    print ("Student " + str(i+1) + ": Hours = " + str(study_hours[i]) + ", Mark = " + str(marks[i]) + " -> [" + status + "]")

passes = 0
for mark in marks:
    if mark >= 60:
        passes += 1

print ("\nNumber of students who passed (60 or above): " + str(passes))
```
